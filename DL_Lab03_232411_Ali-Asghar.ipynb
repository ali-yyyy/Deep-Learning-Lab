Experiment A: Effect of weight's initialization


```python
# Import required libraries
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.initializers import RandomNormal, HeNormal, GlorotNormal
import time # Import the time module

# Define XOR dataset
X = np.array([[0,0],[0,1],[1,0],[1,1]]) # Input combinations
y = np.array([[0],[1],[1],[0]]) # Expected XOR outputs

# Define a simple MLP model
model = Sequential() # Create a sequential neural network container

# Add hidden layer with He initialization
model.add(Dense(3, input_dim=2, activation='relu', kernel_initializer=HeNormal())) # HeNormal suited for ReLU
model.add(Dense(1, activation='sigmoid', kernel_initializer=HeNormal())) # Sigmoid output layer

# Compile the model
model.compile(optimizer='sgd', loss='binary_crossentropy', metrics=['accuracy']) # SGD used for clarity

# Train the model
print('\n--- Training with HeNormal Initialization ---')
start_time = time.time() # Record start time
model.fit(X, y, epochs=100, verbose=1) # 100 epochs for visible convergence trend
end_time = time.time() # Record end time
print(f"Training with HeNormal took: {end_time - start_time:.4f} seconds")
```

 Experiment A: Effect of weight's initialization (RandomNormal)

```python
print('\n--- Training with RandomNormal Initialization ---')

# Define a simple MLP model with RandomNormal initialization
model_random = Sequential()

# Add hidden layer with RandomNormal initialization
model_random.add(Dense(3, input_dim=2, activation="sigmoid", kernel_initializer=RandomNormal())) # Used sigmoid for random initialization to check
model_random.add(Dense(1, kernel_initializer=RandomNormal())) # Activation left blank as per request

# Compile the model
model_random.compile(optimizer='sgd', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model
start_time = time.time() # Record start time
model_random.fit(X, y, epochs=100, verbose=1)
end_time = time.time() # Record end time
print(f"Training with RandomNormal took: {end_time - start_time:.4f} seconds")
```

 Experiment A: Effect of weight's initialization (GlorotNormal/Xavier)

```python
print('\n--- Training with GlorotNormal (Xavier) Initialization ---')

# Define a simple MLP model with GlorotNormal initialization
model_glorot = Sequential()

# Add hidden layer with GlorotNormal initialization
model_glorot.add(Dense(3, input_dim=2, activation="tanh", kernel_initializer=GlorotNormal())) # Activation left blank as per request
model_glorot.add(Dense(1, kernel_initializer=GlorotNormal())) # Activation left blank as per request

# Compile the model
model_glorot.compile(optimizer='sgd', loss='binary_crossentropy', metrics=['accuracy'])

# Train the model
start_time = time.time() # Record start time
model_glorot.fit(X, y, epochs=100, verbose=1)
end_time = time.time() # Record end time
print(f"Training with GlorotNormal took: {end_time - start_time:.4f} seconds")
```

Experiment B: Effect of Activation Function

Part 1: Sigmoid

Here, I'll use HE normal initialization, and then change activation functions


```python
activations = ['sigmoid', 'tanh', 'relu']
for act in activations:
    print(f"\nTesting Activation: {act}")

    model = Sequential()
    model.add(Dense(3, input_dim=2, activation=act, kernel_initializer='he_normal')) #Activation varies here
    model.add(Dense(1, activation='sigmoid'))

    model.compile(optimizer='sgd', loss='binary_crossentropy', metrics=['accuracy'])
    model.fit(X, y, epochs=100, verbose=0)

    loss, acc = model.evaluate(X, y, verbose=0)
    print(f"Accuracy with {act}: {acc:.4f}")
```

ReLU is the most suited for HE initialization, as the range is tailored for this function.

Experiment C – Optimizer Comparison

```python
from tensorflow.keras.optimizers import SGD, RMSprop, Adam

optimizers = {'SGD': SGD(learning_rate=0.1), 'RMSProp':
RMSprop(learning_rate=0.01), 'Adam': Adam(learning_rate=0.01)}

for name, opt in optimizers.items():
    print(f"\nOptimizer: {name}")

    model = Sequential()
    model.add(Dense(3, input_dim=2, activation='relu', kernel_initializer='he_normal'))
    model.add(Dense(1, activation='sigmoid'))

    model.compile(optimizer=opt, loss='binary_crossentropy', metrics=['accuracy'])
    model.fit(X, y, epochs=100, verbose=0)

    loss, acc = model.evaluate(X, y, verbose=0)
    print(f"{name} Accuracy: {acc:.4f}")
```

Experiment D – Learning Rate Effect

```python
rates = [0.01, 0.1, 0.5]

for rate in rates:
    print(f"\nLearning Rate = {rate}")

    model = Sequential()
    model.add(Dense(3, input_dim=2, activation='relu', kernel_initializer='he_normal'))
    model.add(Dense(1, activation='sigmoid'))

    opt = SGD(learning_rate=rate)
    model.compile(optimizer=opt, loss='binary_crossentropy', metrics=['accuracy'])
    model.fit(X, y, epochs=100, verbose=0)

    loss, acc = model.evaluate(X, y, verbose=0)
    print(f"Accuracy: {acc:.4f}")
```

```python
# Import libraries for dataset, preprocessing, and models
import numpy as np # Numerical operations
from sklearn.datasets import load_iris # Iris dataset
from sklearn.model_selection import train_test_split # Train/test split utility
from sklearn.preprocessing import StandardScaler # Feature standardization
from tensorflow.keras.models import Sequential # Keras model container
from tensorflow.keras.layers import Dense # Dense layers for MLP
from tensorflow.keras.optimizers import SGD, RMSprop, Adam# Optimizers to compare
import time # For optional timing (if desired)

# Load Iris dataset and convert to binary classification (Setosa vs Non-Setosa)
iris = load_iris() # Load dataset object
X_full = iris.data # Feature matrix (150, 4)

y_full = (iris.target != 0).astype(int) # Binary label: 0 if Setosa, 1 otherwise

# Standardize features (zero mean, unit variance) for stable training
scaler = StandardScaler() # Create scaler instance
X_scaled = scaler.fit_transform(X_full) # Fit on all data and transform

# Train-test split (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_full, test_size=0.2,
random_state=42)

# Function to build and evaluate model with specified optimizer
def evaluate_optimizer(optimizer, optimizer_name):
# Build a small neural network
model = Sequential() # Initialize sequential model
model.add(Dense(32, activation='relu', input_dim=X_train.shape[1])) # Hidden layer
with 32 units
model.add(Dense(16, activation='relu')) # Second hidden layer with 16 units
model.add(Dense(1, activation='sigmoid')) # Output layer for binary classification

# Compile with given optimizer and binary crossentropy loss
model.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['accuracy'])

# Train the model; verbose=0 to silence per-epoch output
start = time.time() # Record start time (optional)
model.fit(X_train, y_train, epochs=80, batch_size=16, verbose=0)
elapsed = time.time() - start # Compute elapsed time (optional)

# Evaluate on test set and print results including time
loss, acc = model.evaluate(X_test, y_test, verbose=0) # Evaluate without extra logging
print(f"{optimizer_name:10s} -> Test Accuracy: {acc:.4f}, Time(s): {elapsed:.2f}")

# Evaluate standard SGD (no momentum), RMSprop, Adam, SGD with momentum
evaluate_optimizer(SGD(learning_rate=0.01), 'SGD')
evaluate_optimizer(RMSprop(learning_rate=0.001), 'RMSprop')
evaluate_optimizer(Adam(learning_rate=0.001), 'Adam')
evaluate_optimizer(SGD(learning_rate=0.01, momentum=0.9), 'SGD+Mom')
```
